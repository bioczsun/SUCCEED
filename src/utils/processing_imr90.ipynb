```python
contig_df = pd.read_table("/home/suncz/work/s02/Encode_epigenome/results/preprocessed/1m-epcot/sequences.bed" , sep="\t",header= None)
contig_df.columns = ["chrom","start","end","name"]
```

```python
signal_ls = []
num = 0
for seq,signal in zip(contig_df.values,prediction):
    chr = seq[0]
    start = int(seq[1])
    end = int(seq[2])
    signal = signal[:,4]
    num += 1
    for index,res_start in enumerate(range(start+1024*64,end-1024*64,1024)):
        res_end = res_start + 1024
        signal_ls.append([chr,res_start,res_end,signal[index]])
    print(num,end="\r")
tmp_df = pd.DataFrame(signal_ls,columns=["chr","start","end","signal"])
tmp_df.sort_values(by=["chr","start"],inplace=True)
tmp_df.drop_duplicates(subset=["chr","start","end"],inplace=True,keep="first")
tmp_df.to_csv(f"/home/suncz/work/s02/Encode_epigenome/results/benchmark/EPCOT/dataset/succeed/4-cells/model/csv/logs/version_0/imr90_ctcf_succeed_signal.bedGraph",sep="\t",index=False,header=False)
```

```python
tf_list = open("/home/suncz/work/s02/Encode_epigenome/results/benchmark/EPCOT/dataset/succeed/4-cells/tf.txt","r").readlines()
tf_list = [i.split()[0] for i in tf_list]
```

```python
tf_list_dict = {}
for i,j in enumerate(tf_list):
    tf_list_dict[j] = i
```

```python
df_imr90_select = df_imr90[df_imr90["factors"].isin(tf_list)]
```

```python
test_npy = np.load('/home/suncz/work/s02/Encode_epigenome/results/preprocessed/1m-akita/seqs_cov/792.h5.npy', allow_pickle=True)
```

```python
df_imr90_select = df_imr90_select.drop_duplicates(subset=["factors"])
```

```python
tf_name = []
corr_ls = []
for i in df_imr90_select.iloc:
    factors = i.factors
    name = i.name
    pred_index = tf_list_dict[factors]
    pred = prediction[:,:,pred_index]
    pred_chr10 = pred[contig_df[contig_df["chrom"] == "chr10"].index.values]
    target = np.load(f'/home/suncz/work/s02/Encode_epigenome/results/preprocessed/1m-akita/seqs_cov/{name}.h5.npy', allow_pickle=True)[:,64:-64]
    target_chr10 = target[contig_df[contig_df["chrom"] == "chr10"].index.values]
    corr = np.corrcoef(pred_chr10.flatten(), target_chr10.flatten())[0][1]
    pred_chr21 = pred[contig_df[contig_df["chrom"] == "chr21"].index.values]
    target_chr21 = target[contig_df[contig_df["chrom"] == "chr21"].index.values]
    corr_chr21 = np.corrcoef(pred_chr21.flatten(), target_chr21.flatten())[0][1]
    mean_corr = (corr + corr_chr21) / 2  # Calculating the average correlation
    tf_name.append(factors)
    corr_ls.append(mean_corr)
    print(factors, corr, corr_chr21, mean_corr)
succeed_imr90 = pd.DataFrame({"tf":tf_name, "corr":corr_ls})
succeed_imr90.to_csv('/home/suncz/work/s02/Encode_epigenome/results/benchmark/EPCOT/dataset/succeed/4-cells/model/csv/logs/version_0/imr90_corr.csv', index=False)
```
